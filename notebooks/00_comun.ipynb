````python
import json, os, re, sys, hashlib, warnings
from pathlib import Path, PureWindowsPath

import cv2
import numpy as np
import pandas as pd
import yaml
import matplotlib.pyplot as plt
from scipy.optimize import linear_sum_assignment
from scipy.signal import savgol_filter
from IPython.display import display

def _repo_root(start=Path.cwd()):
    for d in (start, *start.parents):
        if (d / "config.yaml").exists() and (d / "notebooks").is_dir():
            return d
    raise FileNotFoundError("No encuentro config.yaml: abre el notebook desde la carpeta del repositorio")

REPO = _repo_root()
CFG = yaml.safe_load((REPO / "config.yaml").read_text(encoding="utf-8"))
ZONA = (REPO / CFG["zona_trabajo"]).resolve()
P = {k: (ZONA / v) for k, v in CFG["rutas"].items()}
P.update(repo=REPO, zona=ZONA, dd=REPO / "datos_derivados", resultados=REPO / "resultados", modelos_repo=REPO / "modelos")
MODO = os.environ.get("COLIBRI_MODO", CFG["modo"])
assert MODO in ("artefactos", "recalcular"), MODO
SEED = CFG["semilla"]
TOL = CFG["tolerancia"]
DETECTOR = REPO / CFG["vigente"]["detector"]
DETECTOR_ANTERIOR = P["modelos"] / CFG["vigente"]["detector_anterior"]
RUNS = P["runs"]
for k in ("runs", "frames"):
    P[k].mkdir(parents=True, exist_ok=True)

def cargar_env(path=REPO / ".env"):
    """Carga KEY=valor de .env en os.environ sin mostrar nada. Devuelve los nombres cargados (no los valores)."""
    if not path.exists():
        return []
    names = []
    for line in path.read_text(encoding="utf-8").splitlines():
        if "=" in line and not line.lstrip().startswith("#"):
            k, v = line.split("=", 1)
            if v.strip():
                os.environ.setdefault(k.strip(), v.strip().strip('"').strip("'")); names.append(k.strip())
    return names

_env = cargar_env()
print(f"repo {REPO} | zona de trabajo {ZONA} | MODO = {MODO} | .env: {', '.join(_env) or 'no hay'}")
````

## 2. Fuentes: los 12 videos

`datos_derivados/fuentes.csv` es el inventario: id, archivo original, carpeta (`vv`, `train`, `test`), origen y cifras
esperadas (fps y frames). Los videos **no se redistribuyen** (Adobe Stock y YouTube); el notebook 01 explica cómo
conseguirlos y comprueba que coinciden. `video_id` es el nombre que usan todos los CSV del proyecto.

````python
FUENTES = pd.read_csv(P["dd"] / "fuentes.csv")

def slug(stem, maxlen=48):
    s = re.sub(r"[^a-z0-9]+", "_", stem.lower()).strip("_")
    return s[:maxlen].rstrip("_")

def video_path(video_id) -> Path:
    r = FUENTES.set_index("video_id").loc[video_id]
    return P["videos"] / r.carpeta / r.archivo

def corto(video_id):
    """Nombre corto (DoJ1, Gu30, i5bE, J4PK, A129...) a partir del video_id o de cualquier texto que lo contenga."""
    for r in FUENTES.itertuples():
        if r.video_id in video_id or str(r.id_origen).lower() in video_id.lower():
            return r.corto
    return video_id

def videos(roles=("vv", "train", "test"), solo_ajuste_union=False, omitir_faltantes=False) -> dict:
    """Equivale a list_videos() de union.py: {video_id: dict(path, rol, W, H, fps)} leyendo cada video.
    Con omitir_faltantes=True, los videos que no estén se saltan (y se listan en FALTAN) en vez de dar error."""
    global FALTAN
    out, FALTAN = {}, []
    for r in FUENTES[FUENTES.rol.isin(roles)].itertuples():
        if solo_ajuste_union and r.rol == "train" and not r.en_ajuste_union:
            continue
        v = dict(path=video_path(r.video_id), rol=r.rol)
        cap = cv2.VideoCapture(str(v["path"]))
        if not cap.isOpened() and omitir_faltantes:
            FALTAN.append(r.corto); continue
        if not cap.isOpened():
            raise FileNotFoundError(f"Falta el video {v['path']} (ver notebook 01, sección 1)")
        v.update(W=int(cap.get(3)), H=int(cap.get(4)), fps=cap.get(cv2.CAP_PROP_FPS)); cap.release()
        out[r.video_id] = v
    return out
````

## 3. Etiquetas y verdad de terreno

Las etiquetas de `humming` (nuestro proyecto de Roboflow) están **congeladas en el repo**, en
`datos_derivados/humming/etiquetas_{v1,v2}/{train,valid,test}/<video_id>_fNNNNN.txt`. Los manifiestos
(`split_manifest_v1.csv`, `split_manifest_v2.csv`) conservan las rutas de Windows con las que se crearon; aquí se
traducen a la estructura nueva por el nombre del archivo:

- etiqueta → `datos_derivados/humming/etiquetas_<v>/<split>/<stem>.txt` (repo)
- imagen → `<frames>/humming_<v>/<split>/images/<stem>.jpg` (zona de trabajo, la regenera el notebook 01)

GT v1 = etiquetas de la versión 1 de Roboflow (S4, S5). GT v2 = versión corregida (S5, D-10).

````python
HUMMING = P["dd"] / "humming"
VV = P["dd"] / "vv"

def yolo_boxes(text, W, H):
    out = []
    for l in text.splitlines():
        if l.strip():
            _, cx, cy, w, h = map(float, l.split())
            out.append([(cx - w / 2) * W, (cy - h / 2) * H, (cx + w / 2) * W, (cy + h / 2) * H])
    return out

def stem_de(ruta):
    return PureWindowsPath(ruta).stem

def label_path(ruta, version, split):
    return HUMMING / f"etiquetas_{version}" / split / f"{stem_de(ruta)}.txt"

def image_path(ruta, version, split):
    return P["frames"] / f"humming_{version}" / split / "images" / f"{stem_de(ruta)}.jpg"

def cargar_manifiesto(version):
    """split_manifest de la versión: ruta (original), split, origen (nuevo|vv), video_id, frame, split_roboflow, cajas, rf_file."""
    m = pd.read_csv(HUMMING / f"split_manifest_{version}.csv")
    m["etiqueta"] = [label_path(r.ruta, version, r.split) for r in m.itertuples()]
    m["imagen"] = [image_path(r.ruta, version, r.split) for r in m.itertuples()]
    return m

def cargar_gt_test(version, vids):
    """GT de los frames test_ en píxeles del video (load_test_gt de union_gt2.py y load_gt de tracking_test.py)."""
    sm = cargar_manifiesto(version)
    sm = sm[(sm.split == "test") & (sm.origen == "nuevo")]
    rows = []
    for r in sm.itertuples():
        v = vids[r.video_id]
        rows.append(dict(video_id=r.video_id, frame=int(r.frame), boxes=yolo_boxes(r.etiqueta.read_text(), v["W"], v["H"])))
    GT = pd.DataFrame(rows).sort_values(["video_id", "frame"]).reset_index(drop=True)
    GT["n_gt"] = GT.boxes.str.len()
    return GT

def cargar_gt_union(vids, version="v1"):
    """load_gt de union.py: vv (todas sus etiquetas de Roboflow menos ignorar_frames) + frames de train_/test_ del manifiesto."""
    import zipfile
    rows = []
    with zipfile.ZipFile(VV / "prueba_etiquetas_corregidas.zip") as z:
        ignore = set(z.read("ignorar_frames.txt").decode().split())
    sel = cargar_manifiesto(version)
    sel_vv = {(r.video_id, int(r.frame)): r.split for r in sel[sel.origen == "vv"].itertuples()}
    for lab in sorted((HUMMING / f"roboflow_{version}_vv").glob("vv*.txt")):
        stem = lab.stem.split("_jpg.rf.")[0]
        vid, f = stem.rsplit("_f", 1)
        f = int(f)
        if stem in ignore or vid not in vids:
            continue
        v = vids[vid]
        rows.append(dict(video_id=vid, frame=f, boxes=yolo_boxes(lab.read_text(), v["W"], v["H"]),
                         split=sel_vv.get((vid, f), "no_usado")))
    for r in sel[sel.origen == "nuevo"].itertuples():
        if r.video_id not in vids:
            continue
        v = vids[r.video_id]
        rows.append(dict(video_id=r.video_id, frame=int(r.frame), boxes=yolo_boxes(r.etiqueta.read_text(), v["W"], v["H"]),
                         split=r.split))
    GT = pd.DataFrame(rows).sort_values(["video_id", "frame"]).reset_index(drop=True)
    GT["n_gt"] = GT.boxes.str.len()
    return GT
````

## 4. Comparar con las cifras del sprint

En modo `artefactos` cada notebook recalcula sus tablas y las compara con las que se guardaron en el sprint
(`resultados/S*/...csv`). `comparar` une por las claves, mide la diferencia absoluta en cada columna numérica y
marca ✓ si todas quedan dentro de la tolerancia de `config.yaml`. **Nunca sobrescribe la referencia.**

````python
def comparar(nuevo: pd.DataFrame, referencia, claves, columnas=None, tol=None, nombre=""):
    ref = pd.read_csv(referencia) if not isinstance(referencia, pd.DataFrame) else referencia
    tol = TOL["metrica"] if tol is None else tol
    columnas = columnas or [c for c in ref.columns if c not in claves and pd.api.types.is_numeric_dtype(ref[c]) and c in nuevo]
    m = ref[claves + columnas].merge(nuevo[claves + columnas], on=claves, how="outer", suffixes=("_ref", "_nuevo"), indicator=True)
    falta = m[m._merge != "both"]
    d = pd.DataFrame({c: (m[f"{c}_nuevo"] - m[f"{c}_ref"]).abs() for c in columnas})
    m["max_dif"] = d.max(axis=1)
    ok = falta.empty and bool((m.max_dif.fillna(0) <= tol).all())
    print(f"{'✓' if ok else '✗'} {nombre}: {len(m)} filas, diferencia máxima {m.max_dif.max():.4g} (tolerancia {tol})"
          + (f", {len(falta)} filas sin pareja" if len(falta) else ""))
    if not ok:
        display(m[(m.max_dif > tol) | (m._merge != "both")].head(20))
    return ok
````

## 5. Tracking (S3, S5)

`track_video` corre `model.track(persist=True)` cuadro a cuadro con un modelo nuevo por video (estado limpio).
Configuraciones: `BASE` + `COMBO` sobre los YAML de Ultralytics, siempre con `gmc_method: none` (cámara fija).
`evaluate_tracks` empareja con la GT por Hungarian a IoU 0,5 y supone **un ave por video** (por eso `ids_por_ave`).
`build_trajectory` es la unión «actual» de S3 (30 frames, 1,5×); la que se usa ahora es `union_tuned` (sección 6).

````python
RUNS_TRACKING = RUNS / "tracking"
# Portado de video-val/resultados/tracking_v2/tracking_test.py
IMGSZ, CONF_CALL, IOU_MATCH = 640, 0.1, 0.5
SHORT_S = 0.5  # tracks más cortos que esto se consideran sospechosos de ser cajas espurias
TRACKERS = {"bytetrack": "bytetrack.yaml", "ocsort": "ocsort.yaml"}
BASE = {"bytetrack": {"track_buffer": 60}, "ocsort": {"track_buffer": 60, "use_byte": True}}
COMBO = {"track_buffer": 300, "match_thresh": 0.95, "new_track_thresh": 0.5}

def configs_tracking():
    """Las 5 configuraciones de S5 (CONFIGS de tracking_test.py): nombre -> (pesos, tracker, overrides)."""
    NEW, OLD = DETECTOR, DETECTOR_ANTERIOR
    return {
        "a_nuevo_oc_combo": (NEW, "ocsort", {**BASE["ocsort"], **COMBO}),
        "b_nuevo_bt_combo": (NEW, "bytetrack", {**BASE["bytetrack"], **COMBO}),
        "c_nuevo_oc_base": (NEW, "ocsort", BASE["ocsort"]),
        "d_nuevo_bt_base": (NEW, "bytetrack", BASE["bytetrack"]),
        "e_viejo_oc_combo": (OLD, "ocsort", {**BASE["ocsort"], **COMBO}),   # solo referencia
    }

# Portado de video-val/resultados/tracking_v2/tracking_test.py
DECISION = ["a_nuevo_oc_combo", "b_nuevo_bt_combo", "c_nuevo_oc_base", "d_nuevo_bt_base"]  # e es solo referencia
CONFIGS = configs_tracking()
````

````python
# Portado de video-val/resultados/tracking_v2/tracking_test.py
def tracker_cfg(name: str, cfg_name: str, overrides: dict) -> Path:
    from ultralytics.utils import ROOT as ULTRA_ROOT
    src = ULTRA_ROOT / "cfg" / "trackers" / TRACKERS[name]
    cfg = yaml.safe_load(src.read_text(encoding="utf-8"))
    cfg.update({k: v for k, v in overrides.items() if k in cfg})
    if "gmc_method" in cfg:
        cfg["gmc_method"] = "none"        # cámara fija
    dst = RUNS_TRACKING / "cfg" / f"tracker_{cfg_name}.yaml"
    dst.parent.mkdir(parents=True, exist_ok=True)
    dst.write_text(yaml.safe_dump(cfg, sort_keys=False), encoding="utf-8")
    return dst


def track_video(weights: Path, video: Path, tracker: str, cfg_path: Path, conf=CONF_CALL, imgsz=IMGSZ) -> pd.DataFrame:
    from ultralytics import YOLO
    model = YOLO(str(weights))                  # modelo nuevo por video y configuración: estado limpio
    cap = cv2.VideoCapture(str(video)); fps = cap.get(cv2.CAP_PROP_FPS)
    rows, frame_idx, t0 = [], 0, pd.Timestamp.now()
    while True:
        t_ms = cap.get(cv2.CAP_PROP_POS_MSEC)
        ok, frame = cap.read()
        if not ok:
            break
        r = model.track(frame, persist=True, tracker=str(cfg_path), conf=conf, imgsz=imgsz, verbose=False)[0]
        if r.boxes is not None and r.boxes.is_track:
            for (cx, cy, w, h), tid, c in zip(r.boxes.xywh.cpu().numpy(), r.boxes.id.int().cpu().tolist(),
                                             r.boxes.conf.cpu().numpy()):
                rows.append(dict(frame=frame_idx, t_ms=t_ms, track_id=tid, cx_px=cx, cy_px=cy,
                                 w_px=w, h_px=h, conf=float(c)))
        frame_idx += 1
    cap.release()
    secs = (pd.Timestamp.now() - t0).total_seconds()
    df = pd.DataFrame(rows, columns=["frame", "t_ms", "track_id", "cx_px", "cy_px", "w_px", "h_px", "conf"])
    df.attrs.update({"fps_declared": fps, "n_frames": frame_idx, "tracker": tracker, "video": str(video),
                     "weights": str(weights), "cfg": str(cfg_path), "fps_proceso": round(frame_idx / max(secs, 1e-6), 1)})
    return df
````

````python
# Portado de video-val/resultados/tracking_v2/tracking_test.py
def iou_matrix(a: np.ndarray, b: np.ndarray) -> np.ndarray:
    if len(a) == 0 or len(b) == 0:
        return np.zeros((len(a), len(b)))
    tl = np.maximum(a[:, None, :2], b[None, :, :2]); br = np.minimum(a[:, None, 2:], b[None, :, 2:])
    inter = np.prod(np.clip(br - tl, 0, None), axis=2)
    area = lambda x: (x[:, 2] - x[:, 0]) * (x[:, 3] - x[:, 1])
    return inter / (area(a)[:, None] + area(b)[None, :] - inter)


def to_xyxy(df: pd.DataFrame) -> np.ndarray:
    return np.c_[df.cx_px - df.w_px / 2, df.cy_px - df.h_px / 2, df.cx_px + df.w_px / 2, df.cy_px + df.h_px / 2]


def evaluate_tracks(GT: pd.DataFrame, vid: str, tr: pd.DataFrame) -> tuple[dict, pd.DataFrame]:
    g = GT[GT.video_id == vid]
    by_frame = dict(tuple(tr.groupby("frame")))
    per = []
    for r in g.itertuples():
        p = by_frame.get(r.frame, tr.iloc[:0])
        gt = np.array(r.boxes, dtype=float).reshape(-1, 4)
        iou = iou_matrix(gt, to_xyxy(p))
        matched = []
        if iou.size:
            gi, pi = linear_sum_assignment(-iou)
            matched = [(i, j) for i, j in zip(gi, pi) if iou[i, j] >= IOU_MATCH]
        ids = p.track_id.to_numpy()
        per.append({"frame": r.frame, "n_gt": len(gt), "n_pred": len(p), "tp": len(matched),
                    "id": int(ids[matched[0][1]]) if matched else None,           # un ave por video
                    "iou": float(iou[matched[0]]) if matched else np.nan})
    per = pd.DataFrame(per)
    seq = per.id.dropna().astype(int).to_numpy()
    idsw = int((seq[1:] != seq[:-1]).sum()) if len(seq) else 0
    n_gt, n_pred, tp = per.n_gt.sum(), per.n_pred.sum(), per.tp.sum()
    idtp = int(pd.Series(seq).value_counts().iloc[0]) if len(seq) else 0
    m = {"video_id": vid, "tracker": tr.attrs["tracker"], "n_gt": int(n_gt), "TP": int(tp), "FN": int(n_gt - tp),
         "FP": int(n_pred - tp), "IDSW": idsw, "ids_por_ave": int(len(set(seq))),
         "ids_totales": int(tr.track_id.nunique()),
         "recall": tp / max(n_gt, 1), "precision": tp / max(n_pred, 1),
         "MOTA": 1 - (n_gt - tp + n_pred - tp + idsw) / max(n_gt, 1),
         "IDF1": 2 * idtp / max(n_gt + n_pred, 1), "IoU_medio": per.iou.mean(),
         "fps_proceso": tr.attrs.get("fps_proceso"), "idtp": idtp}
    return m, per
````

````python
# Portado de video-val/resultados/tracking_v2/tracking_test.py
INTERP_MAX_GAP = 10
JOIN_MAX_GAP = 30
JOIN_MAX_DIST = 1.5
MIN_FRAG_LEN = 15

# Portado de video-val/resultados/tracking_v2/tracking_test.py
def build_trajectory(tr: pd.DataFrame) -> tuple[pd.DataFrame, dict]:
    fps = tr.attrs["fps_declared"]
    if tr.empty:
        return tr.iloc[:0], {"ids_unidos": [], "ids_descartados": []}
    span = tr.groupby("track_id").frame.agg(["min", "max", "size"]).sort_values("size", ascending=False)
    main = [span.index[0]]
    discarded = []
    for tid in span.index[1:]:
        s = span.loc[tid]
        used = tr[tr.track_id.isin(main)]
        if ((used.frame >= s["min"]) & (used.frame <= s["max"])).any():
            discarded.append(int(tid)); continue
        before, after = used[used.frame < s["min"]], used[used.frame > s["max"]]
        frag = tr[tr.track_id == tid].sort_values("frame")
        ok, near = False, False
        for ref, ends, gap in ((before.sort_values("frame").tail(1), frag.head(1), s["min"] - before.frame.max() if len(before) else np.inf),
                               (after.sort_values("frame").head(1), frag.tail(1), after.frame.min() - s["max"] if len(after) else np.inf)):
            if len(ref) and gap <= JOIN_MAX_GAP:
                near = True
                size = (ref.w_px.iloc[0] + ref.h_px.iloc[0]) / 2
                d = np.hypot(ref.cx_px.iloc[0] - ends.cx_px.iloc[0], ref.cy_px.iloc[0] - ends.cy_px.iloc[0])
                ok |= d <= JOIN_MAX_DIST * size
        if not near:
            ok = s["size"] >= MIN_FRAG_LEN
        (main if ok else discarded).append(int(tid))
    t = tr[tr.track_id.isin(main)].sort_values("frame").drop_duplicates("frame")
    t = t.set_index("frame")[["cx_px", "cy_px", "w_px", "h_px", "conf"]]
    full = t.reindex(range(t.index.min(), t.index.max() + 1))
    missing = full.cx_px.isna()
    run_id = (~missing).cumsum()
    run_len = missing.groupby(run_id).transform("sum")
    fill = missing & (run_len <= INTERP_MAX_GAP)
    interp = full.interpolate(limit_area="inside")
    full.loc[fill, ["cx_px", "cy_px", "w_px", "h_px"]] = interp.loc[fill, ["cx_px", "cy_px", "w_px", "h_px"]]
    full["interpolado"] = fill
    full = full[full.cx_px.notna()].copy()
    full["segmento"] = (np.diff(full.index, prepend=full.index[0]) > 1).cumsum()
    full["t_s"] = full.index / fps
    v = np.full(len(full), np.nan)
    for _, seg in full.groupby("segmento"):
        if len(seg) >= 7:
            w = min(11, len(seg) // 2 * 2 - 1)
            vx = savgol_filter(seg.cx_px, w, 2, deriv=1) * fps
            vy = savgol_filter(seg.cy_px, w, 2, deriv=1) * fps
            v[full.index.get_indexer(seg.index)] = np.hypot(vx, vy)
    full["v_px_s"] = v
    full = full.reset_index().rename(columns={"index": "frame"})
    return full[["frame", "t_s", "cx_px", "cy_px", "w_px", "h_px", "conf", "interpolado", "segmento", "v_px_s"]], \
        {"ids_unidos": [int(x) for x in main], "ids_descartados": discarded}


def short_tracks(tr: pd.DataFrame, GT_v: pd.DataFrame) -> pd.DataFrame:
    fps = tr.attrs["fps_declared"]
    g = tr.groupby("track_id").agg(ini=("frame", "min"), fin=("frame", "max"), n=("frame", "size"),
                                   cx=("cx_px", "mean"), cy=("cy_px", "mean"), w=("w_px", "mean"),
                                   h=("h_px", "mean"), conf=("conf", "mean"))
    g["dur_s"] = (g.fin - g.ini + 1) / fps
    g = g[g.dur_s < SHORT_S].reset_index()
    # ¿había ave etiquetada en el frame GT más cercano?
    gf = GT_v.frame.to_numpy()
    g["ave_en_gt_cercano"] = [bool(GT_v.n_gt.iloc[np.abs(gf - (a + b) / 2).argmin()]) if len(gf) else None
                              for a, b in zip(g.ini, g.fin)]
    g["t_ini_s"] = g.ini / fps
    return g


def draw_traj(video: Path, tr, traj, GT_v, title, out):
    cap = cv2.VideoCapture(str(video))
    mid = int(traj.frame.median()) if len(traj) else 0
    cap.set(cv2.CAP_PROP_POS_FRAMES, mid); ok, f = cap.read(); cap.release()
    fig, ax = plt.subplots(figsize=(12, 12 * f.shape[0] / f.shape[1]))
    ax.imshow(cv2.cvtColor(f, cv2.COLOR_BGR2RGB), alpha=0.55)
    for tid, g in tr.groupby("track_id"):           # todos los IDs, finos
        ax.plot(g.cx_px, g.cy_px, lw=0.8, alpha=0.7)
    sc = None
    for _, seg in traj.groupby("segmento"):         # trayectoria final, gruesa y coloreada por tiempo
        sc = ax.scatter(seg.cx_px, seg.cy_px, c=seg.t_s, cmap="viridis", s=8, vmin=0, vmax=tr.attrs["n_frames"] / tr.attrs["fps_declared"])
    gx = [(b[0][0] + b[0][2]) / 2 for b in GT_v.boxes if b]; gy = [(b[0][1] + b[0][3]) / 2 for b in GT_v.boxes if b]
    ax.scatter(gx, gy, marker="x", c="red", s=22, label="verdad de terreno")
    ax.set_title(title, fontsize=10); ax.axis("off"); ax.legend(loc="lower left", fontsize=8)
    if sc is not None:
        fig.colorbar(sc, ax=ax, shrink=0.6, label="t (s)")
    fig.savefig(out, dpi=110, bbox_inches="tight"); plt.close(fig)
````

## 6. Unión de fragmentos a nivel de visita (S5)

`union_tuned(tr, max_gap_s, max_dist, cuts)`: recorre las filas aceptadas en orden temporal; dos filas consecutivas
siguen en el mismo segmento si el hueco es ≤ `max_gap_s`, la distancia ≤ `max_dist` × lado medio de la caja y no hay
corte de escena entre ellas. Los huecos dentro de un segmento se interpolan (**puente**). Vigente (D-10): 1 s y 2,5×.

`visit_metrics`: una **visita** es una racha de frames GT con ave. `IDF1_visita` empareja visitas con segmentos;
un **puente falso** es un segmento que cubre dos visitas. `bridges` cuenta los puentes **largos** (> ¼ del lado corto).

````python
# Portado de video-val/resultados/tracking_v2/union/union.py
IOU_MATCH = 0.5
MIN_SEG_S = 0.5
CUT_THR = 0.5  # correlación de histograma HS entre frames consecutivos por debajo de la cual hay corte de escena

# Portado de video-val/resultados/tracking_v2/union/union.py
def scene_cuts(path: Path) -> list:
    f = RUNS / "union" / "cortes" / f"{path.stem[:60]}.json"
    if f.exists():
        return json.loads(f.read_text())
    cap = cv2.VideoCapture(str(path)); prev, cuts, i = None, [], 0
    while True:
        ok = cap.grab()
        if not ok:
            break
        ok, fr = cap.retrieve()
        hsv = cv2.cvtColor(cv2.resize(fr, (160, 90)), cv2.COLOR_BGR2HSV)
        h = cv2.normalize(cv2.calcHist([hsv], [0, 1], None, [32, 32], [0, 180, 0, 256]), None).flatten()
        if prev is not None and cv2.compareHist(prev, h, cv2.HISTCMP_CORREL) < CUT_THR:
            cuts.append(i)  # el corte ocurre entre i-1 e i
        prev, i = h, i + 1
    cap.release()
    f.parent.mkdir(parents=True, exist_ok=True); f.write_text(json.dumps(cuts))
    return cuts
````

````python
# Portado de video-val/resultados/tracking_v2/union/union.py
def accepted_rows(tr):
    """Aceptación de fragmentos del notebook: el mayor primero; fuera los que se solapan en el tiempo con uno aceptado."""
    span = tr.groupby("track_id").frame.agg(["min", "max", "size"]).sort_values("size", ascending=False)
    acc = []
    for tid, s in span.iterrows():
        if any(not (s["max"] < span.loc[a, "min"] or s["min"] > span.loc[a, "max"]) for a in acc):
            continue
        acc.append(tid)
    return tr[tr.track_id.isin(acc)].sort_values("frame").drop_duplicates("frame").reset_index(drop=True)


def union_tuned(tr, max_gap_s, max_dist, cuts):
    fps = tr.attrs["fps_declared"]
    if tr.empty:
        return pd.DataFrame(columns=["frame", "t_s", "cx_px", "cy_px", "w_px", "h_px", "conf", "track_id",
                                     "interpolado", "segment_id"])
    a = accepted_rows(tr)
    cuts = np.array(cuts)
    seg = np.zeros(len(a), int)
    for i in range(1, len(a)):
        p, q = a.iloc[i - 1], a.iloc[i]
        gap = (q.frame - p.frame) / fps
        size = (p.w_px + p.h_px) / 2
        dist = np.hypot(q.cx_px - p.cx_px, q.cy_px - p.cy_px)
        cut = ((cuts > p.frame) & (cuts <= q.frame)).any() if len(cuts) else False
        seg[i] = seg[i - 1] + (not (gap <= max_gap_s + 1e-9 and dist <= max_dist * size and not cut))
    a["segment_id"] = seg
    parts = []
    for sid, g in a.groupby("segment_id"):
        if len(g) < MIN_SEG_S * fps:
            continue
        full = g.set_index("frame").reindex(range(g.frame.min(), g.frame.max() + 1))
        miss = full.cx_px.isna()
        full[["cx_px", "cy_px", "w_px", "h_px"]] = full[["cx_px", "cy_px", "w_px", "h_px"]].interpolate()
        full["interpolado"] = miss
        full["segment_id"] = sid
        parts.append(full.reset_index().rename(columns={"index": "frame"}))
    if not parts:
        return union_tuned(tr.iloc[:0], max_gap_s, max_dist, cuts)
    out = pd.concat(parts, ignore_index=True)
    out["segment_id"] = out.segment_id.rank(method="dense").astype(int) - 1
    out["t_s"] = out.frame / fps
    return out[["frame", "t_s", "cx_px", "cy_px", "w_px", "h_px", "conf", "track_id", "interpolado", "segment_id"]]


def union_reference(tr):
    traj, _ = build_trajectory(tr)
    return traj.rename(columns={"segmento": "segment_id"})
````

````python
# Portado de video-val/resultados/tracking_v2/union/union.py
def visit_metrics(traj, g):
    """g: frames GT de un video. Frames con 2+ cajas se excluyen (no forman ni separan visitas)."""
    g = g[g.n_gt <= 1].sort_values("frame")
    present = (g.n_gt == 1).to_numpy()
    visit = np.where(present, np.cumsum(np.r_[present[0], present[1:] & ~present[:-1]]) - 1, -1)
    by_frame = traj.set_index("frame") if len(traj) else None
    match, n_pred = [], 0  # (visita, segmento) por caja GT cubierta
    for (f, boxes), vi in zip(zip(g.frame, g.boxes), visit):
        if by_frame is None or f not in by_frame.index:
            continue
        r = by_frame.loc[f]
        n_pred += 1
        if vi >= 0:
            pb = np.array([[r.cx_px - r.w_px / 2, r.cy_px - r.h_px / 2, r.cx_px + r.w_px / 2, r.cy_px + r.h_px / 2]])
            if iou_matrix(np.array(boxes), pb)[0, 0] >= IOU_MATCH:
                match.append((vi, int(r.segment_id)))
    n_vis, n_box = int(visit.max() + 1) if len(visit) else 0, int(present.sum())
    M = pd.DataFrame(match, columns=["v", "s"])
    seg_per_visit = M.groupby("v").s.nunique().reindex(range(n_vis), fill_value=0)
    bridges = int((M.groupby("s").v.nunique() - 1).clip(lower=0).sum()) if len(M) else 0
    if len(M):
        C = M.value_counts().unstack(fill_value=0)
        ri, ci = linear_sum_assignment(-C.to_numpy())
        idtp = int(C.to_numpy()[ri, ci].sum())
    else:
        idtp = 0
    return dict(visitas=n_vis, cajas=n_box, cubiertas=len(M), pred_en_gt=n_pred, idtp=idtp,
                seg_por_visita=float(seg_per_visit[seg_per_visit > 0].mean()) if (seg_per_visit > 0).any() else 0.0,
                visitas_sin_segmento=int((seg_per_visit == 0).sum()), puentes_falsos=bridges,
                cobertura=len(M) / max(n_box, 1), IDF1_visita=2 * idtp / max(n_box + n_pred, 1),
                segmentos=int(traj.segment_id.nunique()) if len(traj) else 0)


def aggregate(rows):
    d = pd.DataFrame(rows)
    s = d[["visitas", "cajas", "cubiertas", "pred_en_gt", "idtp", "visitas_sin_segmento", "puentes_falsos", "segmentos"]].sum()
    w = d.visitas - d.visitas_sin_segmento
    return dict(**s.to_dict(), seg_por_visita=float((d.seg_por_visita * w).sum() / max(w.sum(), 1)),
                cobertura=s.cubiertas / max(s.cajas, 1), IDF1_visita=2 * s.idtp / max(s.cajas + s.pred_en_gt, 1))


def draw(video, traj, g, title, out):
    cap = cv2.VideoCapture(str(video))
    mid = int(traj.frame.median()) if len(traj) else 0
    cap.set(cv2.CAP_PROP_POS_FRAMES, mid); ok, f = cap.read(); cap.release()
    fig, ax = plt.subplots(figsize=(12, 12 * f.shape[0] / f.shape[1]))
    ax.imshow(cv2.cvtColor(f, cv2.COLOR_BGR2RGB), alpha=0.5)
    cmap = plt.get_cmap("tab10")
    for sid, s in traj.groupby("segment_id"):
        c = cmap(int(sid) % 10)
        s = s.sort_values("frame")
        real = ~s.interpolado.astype(bool)
        # tramos reales en línea continua, puentes interpolados en discontinua
        brk = (real != real.shift()).cumsum()
        for _, part in s.groupby(brk):
            ax.plot(part.cx_px, part.cy_px, color=c, lw=2, ls="-" if real.loc[part.index[0]] else "--")
        ax.annotate(f"s{sid}", (s.cx_px.iloc[0], s.cy_px.iloc[0]), color=c, fontsize=9, weight="bold")
    gx = [(b[0][0] + b[0][2]) / 2 for b in g.boxes if b]; gy = [(b[0][1] + b[0][3]) / 2 for b in g.boxes if b]
    if gx:   # sin verdad de terreno (videos propios) no se dibuja
        ax.scatter(gx, gy, marker="x", c="red", s=20, label="verdad de terreno")
    ax.plot([], [], "k--", label="puente (interpolado)")
    ax.set_title(title, fontsize=10); ax.axis("off"); ax.legend(loc="lower left", fontsize=8)
    fig.savefig(out, dpi=110, bbox_inches="tight"); plt.close(fig)

# Portado de video-val/resultados/tracking_v2/eval_gt2/union_gt2.py
def bridges(traj, W, H):
    """Tramos interpolados: (segmento, frame inicio, frame fin, longitud px); y el número de puentes largos."""
    out = []
    if traj.empty:
        return out, 0
    t = traj.sort_values("frame").reset_index(drop=True)
    interp = t.interpolado.astype(bool).to_numpy()
    i = 0
    while i < len(t):
        if interp[i]:
            j = i
            while j < len(t) and interp[j] and t.segment_id[j] == t.segment_id[i]:
                j += 1
            a, b = max(i - 1, 0), min(j, len(t) - 1)
            L = float(np.hypot(t.cx_px[b] - t.cx_px[a], t.cy_px[b] - t.cy_px[a]))
            out.append(dict(segment_id=int(t.segment_id[i]), frame_ini=int(t.frame[i]), frame_fin=int(t.frame[j - 1]), long_px=L))
            i = j
        else:
            i += 1
    thr = min(W, H) / 4
    return out, sum(1 for o in out if o["long_px"] > thr)

def union_vigente(tr, cuts):
    u = CFG["vigente"]["union"]
    traj = union_tuned(tr, u["max_gap_s"], u["max_dist"], cuts).copy()
    traj["es_puente"] = traj.interpolado.astype(bool)
    return traj
````

## 7. Siete puntos (S7)

Suavizado Savitzky-Golay por segmento (~0,2 s, grado 2); 7 puntos **en total** por video (D-12) con dos métodos:
uniforme en el tiempo con ave y salientes (Visvalingam-Whyatt). Un punto que cae en un puente se mueve al observado
más cercano. Por los puntos de cada tramo pasa una Catmull-Rom centrípeta; los huecos entre segmentos no se dibujan.

````python
# Portado de video-val/resultados/trayectorias_7pts/trayectorias_7pts.py
N = 7
SMOOTH_S = 0.2
NAMES = {"doj1xa6ksqm": "DoJ1", "gu30nofexla": "Gu30", "i5befvb5mxg": "i5bE", "j4pkgwgjaus": "J4PK"}
METHODS = {"a_uniforme_tiempo": "uniforme en el tiempo", "b_salientes": "salientes (Visvalingam-Whyatt)"}

# Portado de video-val/resultados/trayectorias_7pts/trayectorias_7pts.py
def smooth(xy, fps):
    w = max(5, int(round(SMOOTH_S * fps)) | 1)
    if len(xy) <= w:
        return xy.copy()
    return np.c_[savgol_filter(xy[:, 0], w, 2), savgol_filter(xy[:, 1], w, 2)]


def uniform_time(xy, n=N):
    # el recorrido tiene una fila por frame con ave, así que igual número de filas es igual tiempo con ave
    return np.linspace(0, len(xy) - 1, n).round().astype(int).tolist()


def visvalingam(xy, n=N):
    idx = list(range(len(xy)))
    area = lambda a, b, c: 0.5 * abs((xy[b, 0] - xy[a, 0]) * (xy[c, 1] - xy[a, 1]) - (xy[c, 0] - xy[a, 0]) * (xy[b, 1] - xy[a, 1]))
    A = [np.inf] + [area(idx[i - 1], idx[i], idx[i + 1]) for i in range(1, len(idx) - 1)] + [np.inf]
    while len(idx) > n:
        k = int(np.argmin(A))
        floor = A[k]  # área efectiva: una vecina nunca baja de la del punto eliminado
        del idx[k], A[k]
        for j in (k - 1, k):
            if 0 < j < len(idx) - 1:
                A[j] = max(area(idx[j - 1], idx[j], idx[j + 1]), floor)
    return idx


def fix_bridges(idx, bridge):
    """Mueve a un punto observado los que caen en un puente, sin repetir ni desordenar. Devuelve índices y marcas."""
    obs = np.flatnonzero(~bridge)
    out, moved, used = [], [], set()
    for i in idx:
        was = bool(bridge[i])
        cand = sorted(obs, key=lambda o: abs(o - i)) if was else [i]
        j = next((c for c in cand if c not in used), i)
        used.add(j); out.append(int(j)); moved.append(was)
    order = np.argsort(out)
    return [out[o] for o in order], [moved[o] for o in order]


def catmull_rom(P, per_seg=40, alpha=0.5):
    P = np.asarray(P, float)
    if len(P) == 1:
        return P.copy()
    Q = np.vstack([2 * P[0] - P[1], P, 2 * P[-1] - P[-2]])
    out = []
    for i in range(len(Q) - 3):
        p0, p1, p2, p3 = Q[i:i + 4]
        t0 = 0.0
        t1 = t0 + max(np.linalg.norm(p1 - p0), 1e-6) ** alpha
        t2 = t1 + max(np.linalg.norm(p2 - p1), 1e-6) ** alpha
        t3 = t2 + max(np.linalg.norm(p3 - p2), 1e-6) ** alpha
        t = np.linspace(t1, t2, per_seg, endpoint=i == len(Q) - 4)[:, None]
        A1 = (t1 - t) / (t1 - t0) * p0 + (t - t0) / (t1 - t0) * p1
        A2 = (t2 - t) / (t2 - t1) * p1 + (t - t1) / (t2 - t1) * p2
        A3 = (t3 - t) / (t3 - t2) * p2 + (t - t2) / (t3 - t2) * p3
        B1 = (t2 - t) / (t2 - t0) * A1 + (t - t0) / (t2 - t0) * A2
        B2 = (t3 - t) / (t3 - t1) * A2 + (t - t1) / (t3 - t1) * A3
        out.append((t2 - t) / (t2 - t1) * B1 + (t - t1) / (t2 - t1) * B2)
    return np.vstack(out)


def curves_by_stretch(P, seg_ids):
    """Una curva por tramo continuo: se corta cuando dos puntos consecutivos están en segmentos distintos."""
    cuts = np.flatnonzero(np.diff(seg_ids) != 0) + 1
    return [catmull_rom(p) for p in np.split(P, cuts)]


def shape_error(dense, curve):
    d = np.sqrt(((dense[:, None, :] - curve[None, :, :]) ** 2).sum(-1)).min(1)
    return float(d.mean()), float(d.max())


def gaps_between(sel):
    """Huecos entre puntos consecutivos: (índice del punto anterior, segundos sin ave)."""
    return [(k, float(sel.t_s.iloc[k + 1] - sel.t_s.iloc[k])) for k in range(len(sel) - 1)
            if sel.segment_id.iloc[k] != sel.segment_id.iloc[k + 1]]


def mark_gaps(ax, P, gaps, color="0.35"):
    for k, secs in gaps:
        mid = (P[k] + P[k + 1]) / 2
        ax.annotate(f"hueco {secs:.1f} s", mid, ha="center", va="center", fontsize=9, color=color, style="italic")


def sketch(ax, curves, P, gaps, title=None, numbers=False):
    for c in curves:
        ax.plot(c[:, 0], c[:, 1], color="black", lw=3, solid_capstyle="round")
    ax.scatter(P[:, 0], P[:, 1], s=26, c="black", zorder=3)
    if numbers:
        for i, (x, y) in enumerate(P, 1):
            ax.annotate(str(i), (x, y), xytext=(6, 6), textcoords="offset points", fontsize=10, color="0.25")
        mark_gaps(ax, P, gaps)
    allp = np.vstack(curves)
    m = 0.08 * max(np.ptp(allp[:, 0]), np.ptp(allp[:, 1]), 1)
    ax.set_xlim(allp[:, 0].min() - m, allp[:, 0].max() + m); ax.set_ylim(allp[:, 1].max() + m, allp[:, 1].min() - m)
    ax.set_aspect("equal"); ax.set_facecolor("white"); ax.axis("off")
    if title:
        ax.set_title(title, fontsize=12)


def sketch_norm(ax, curves, P, gaps, W, H, title):
    """Misma escala para todos: el cuadro completo del video, normalizado de 0 a 1 (proporción 16:9)."""
    s = np.array([W, H], float)
    for c in curves:
        ax.plot(c[:, 0] / W, c[:, 1] / H, color="black", lw=3, solid_capstyle="round")
    Pn = P / s
    ax.scatter(Pn[:, 0], Pn[:, 1], s=26, c="black", zorder=3)
    for i, (x, y) in enumerate(Pn, 1):
        ax.annotate(str(i), (x, y), xytext=(6, 6), textcoords="offset points", fontsize=10, color="0.25")
    mark_gaps(ax, Pn, gaps)
    ax.add_patch(plt.Rectangle((0, 0), 1, 1, fill=False, ec="0.6", lw=1))
    ax.set_xlim(-0.01, 1.01); ax.set_ylim(1.01, -0.01); ax.set_aspect(H / W); ax.axis("off")
    ax.set_title(title, fontsize=12)

# Cuerpo por video de main() de trayectorias_7pts.py, separado para usarlo también en la herramienta
def siete_puntos(t, fps):
    """t: trayectoria ordenada por frame. Devuelve el recorrido denso y, por método, índices, puntos, curvas y error."""
    dense = np.vstack([smooth(s[["cx_px", "cy_px"]].to_numpy(), fps) for _, s in t.groupby("segment_id", sort=False)])
    bridge = t.es_puente.astype(bool).to_numpy()
    seg_ids = t.segment_id.to_numpy()
    out = dict(dense=dense, seg_ids=seg_ids, metodos={})
    for m, fn in (("a_uniforme_tiempo", uniform_time), ("b_salientes", visvalingam)):
        idx, moved = fix_bridges(fn(dense), bridge)
        pts = dense[idx]
        sel = t.iloc[idx]
        curves = curves_by_stretch(pts, seg_ids[idx])
        gaps = gaps_between(sel)
        e_mean, e_max = shape_error(dense, np.vstack(curves))
        out["metodos"][m] = dict(idx=idx, moved=moved, P=pts, sel=sel, curves=curves, gaps=gaps, e_mean=e_mean, e_max=e_max,
                                 size=float(np.hypot(*np.ptp(dense, axis=0))))
    return out


def tabla_puntos(r, W, H):
    """CSV de 7 puntos de un método: la columna hueco_antes marca dónde se corta la curva."""
    pts, sel = r["P"], r["sel"]
    hueco = [False] * len(r["idx"])
    for k, _ in r["gaps"]:
        hueco[k + 1] = True
    return pd.DataFrame(dict(punto=range(1, len(r["idx"]) + 1), frame=sel.frame.to_numpy(), t_s=sel.t_s.round(3).to_numpy(),
                             segment_id=sel.segment_id.to_numpy(), x_px=pts[:, 0].round(1), y_px=pts[:, 1].round(1),
                             x_norm=(pts[:, 0] / W).round(4), y_norm=(pts[:, 1] / H).round(4), es_puente_corregido=r["moved"],
                             hueco_antes=hueco))


def lamina_fondo(sp, r, bg, W, H, titulo, salida):
    f1, a1 = plt.subplots(figsize=(12, 12 * H / W))
    a1.imshow(bg, alpha=0.55)
    for sid in np.unique(sp["seg_ids"]):
        d = sp["dense"][sp["seg_ids"] == sid]
        a1.plot(d[:, 0], d[:, 1], color="0.3", lw=1)
    for c in r["curves"]:
        a1.plot(c[:, 0], c[:, 1], color="crimson", lw=2.5)
    pts = r["P"]
    a1.scatter(pts[:, 0], pts[:, 1], s=60, c="crimson", edgecolors="white", zorder=5)
    for k, (x, y) in enumerate(pts, 1):
        a1.annotate(str(k), (x, y), xytext=(6, 6), textcoords="offset points", fontsize=11, weight="bold",
                    bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.85))
    mark_gaps(a1, pts, r["gaps"], color="black")
    a1.set_title(titulo, fontsize=11); a1.axis("off")
    f1.savefig(salida, dpi=110, bbox_inches="tight"); plt.close(f1)


def lamina_trazo(r, salida):
    f2, a2 = plt.subplots(figsize=(10, 10))
    sketch(a2, r["curves"], r["P"], r["gaps"])
    f2.savefig(salida, dpi=110, bbox_inches="tight", facecolor="white"); plt.close(f2)
````

## 8. Profundidad (S8)

Todos los modelos de profundidad entran por la misma interfaz: `map(frame_idx, frame, flip, secondary)` devuelve un
mapa alineado con el frame original. `yolo26` es profundidad (mayor = más lejos); `vda` es **disparidad** (mayor = más
cerca) y lee los mapas que calculó el notebook 05. `z_norm = (z_ave − z_flor)/(z_fondo − z_flor)`: 0 = flor,
1 = fondo, negativo = delante, en los dos dominios (D-14). La magnitud no es comparable entre dominios.

````python
# Portado de video-val/resultados/profundidad/siete_puntos/comun_profundidad.py
def read_frame(cap, f):
    cap.set(cv2.CAP_PROP_POS_FRAMES, int(f)); ok, fr = cap.read()
    return fr if ok else None


def motion_mask(frame, bg, x0, y0, x1, y1):
    # quita el cambio global de brillo; el umbral se adapta al ruido del anillo alrededor de la caja
    pad = max(10, int(0.5 * max(x1 - x0, y1 - y0)))
    X0, Y0, X1, Y1 = max(0, x0 - pad), max(0, y0 - pad), min(frame.shape[1], x1 + pad), min(frame.shape[0], y1 + pad)
    diff = cv2.absdiff(frame[Y0:Y1, X0:X1], bg[Y0:Y1, X0:X1]).astype(np.float32)
    diff = np.abs(diff - np.median(diff, axis=(0, 1))).max(axis=2)
    ring = np.ones(diff.shape, bool); ring[y0 - Y0:y1 - Y0, x0 - X0:x1 - X0] = False
    thr = max(15.0, float(np.percentile(diff[ring], 95)) + 5) if ring.any() else 15.0
    m = (diff > thr).astype(np.uint8)
    m = cv2.morphologyEx(m, cv2.MORPH_OPEN, np.ones((3, 3), np.uint8))
    return m[y0 - Y0:y1 - Y0, x0 - X0:x1 - X0].astype(bool)
````

````python
# Portado de video-val/resultados/profundidad/siete_puntos/backends_profundidad.py
class Yolo26:
    variants = ("orig", "flip", "secondary")

    def __init__(self, model, secondary="yolo26m-depth.pt", imgsz=640):
        from ultralytics import YOLO
        self.m, self.s, self.imgsz = YOLO(model), YOLO(secondary), imgsz

    def _run(self, m, img):
        return m.predict(img, imgsz=self.imgsz, verbose=False)[0].depth.data.cpu().numpy().astype(np.float32)

    def map(self, frame_idx, frame, flip=False, secondary=False):
        m = self.s if secondary else self.m
        return self._run(m, frame[:, ::-1].copy())[:, ::-1] if flip else self._run(m, frame)


class Yolo26Crop(Yolo26):
    def __init__(self, model, crop, secondary="yolo26m-depth.pt", imgsz=640):
        super().__init__(model, secondary, imgsz)
        self.crop = crop  # (x0, y0, x1, y1) en píxeles, fijo para todo el video

    def map(self, frame_idx, frame, flip=False, secondary=False):
        x0, y0, x1, y1 = self.crop
        out = np.full(frame.shape[:2], np.nan, np.float32)
        out[y0:y1, x0:x1] = super().map(frame_idx, frame[y0:y1, x0:x1], flip, secondary)
        return out


class VDA:
    variants = ("orig", "flip")

    def __init__(self, cache_dir, name, shape):
        self.o = np.load(Path(cache_dir) / f"{name}_orig.npz")["maps"]
        self.f = np.load(Path(cache_dir) / f"{name}_flip.npz")["maps"]
        self.H, self.W = shape

    def map(self, frame_idx, frame, flip=False, secondary=False):
        src = self.f if flip else self.o
        return cv2.resize(src[int(frame_idx)].astype(np.float32), (self.W, self.H), interpolation=cv2.INTER_LINEAR)


def make_backend(kind, model, name, shape, crop=None, cache_dir=None):
    if kind == "yolo26":
        return Yolo26(model + ".pt")
    if kind == "yolo26_crop":
        return Yolo26Crop(model + ".pt", crop)
    if kind == "vda":
        return VDA(cache_dir, name, shape)
    raise ValueError(kind)
````

````python
# Portado del notebook colibri_3d_7puntos (S8); antes vivía en 06. Lo usan 06 y la herramienta
N_BG, MIN_MASK_PX = 60, 30

def rect_px(r, W, H):
    return int(r[0] * W), int(r[1] * H), int(r[2] * W), int(r[3] * H)

def background(cap, n):
    fs = [read_frame(cap, i) for i in np.linspace(int(0.03 * n), n - 1, N_BG).astype(int)]
    fs = [f for f in fs if f is not None and cv2.cvtColor(f, cv2.COLOR_BGR2GRAY).mean() >= 5]
    return np.median(np.stack(fs), axis=0).astype(np.uint8)

def median_outside(d, rect, box):
    x0, y0, x1, y1 = rect; m = np.ones((y1 - y0, x1 - x0), bool)
    ax0, ay0, ax1, ay1 = max(box[0], x0), max(box[1], y0), min(box[2], x1), min(box[3], y1)
    if ax1 > ax0 and ay1 > ay0:
        m[ay0 - y0:ay1 - y0, ax0 - x0:ax1 - x0] = False
    return float(np.nanmedian(d[y0:y1, x0:x1][m])) if m.mean() > 0.2 else float(np.nanmedian(d[y0:y1, x0:x1]))

def read_point(d, box, mask, ref, far):
    x0, y0, x1, y1 = box; b = d[y0:y1, x0:x1]; h, w = b.shape
    use = int(mask.sum()) >= MIN_MASK_PX
    z = float(np.nanmedian(b[mask])) if use else float(np.nanmedian(b[h // 4:h - h // 4 or None, w // 4:w - w // 4 or None]))
    fl, fo = median_outside(d, ref, box), median_outside(d, far, box)
    return dict(z_ave=z, z_flor=fl, z_fondo=fo, z_norm=(z - fl) / (fo - fl), den=fo - fl, metodo_z="mask_med" if use else "recorte")

def box_of(r, W, H):
    return (max(0, int(r.cx_px - r.w_px / 2)), max(0, int(r.cy_px - r.h_px / 2)), min(W, int(r.cx_px + r.w_px / 2) + 1), min(H, int(r.cy_px + r.h_px / 2) + 1))


def leer_puntos_z(PT, t, cap, be, bg, ref, far, W, H):
    """Lee z en cada punto de PT: original + volteado (TTA) y vecinos a ±2 frames. t: trayectoria indexada por frame."""
    def at(f, tta):
        fr = read_frame(cap, f); box = box_of(t.loc[f], W, H); mask = motion_mask(fr, bg, *box)
        o = read_point(be.map(f, fr), box, mask, ref, far); o["n_px"] = int(mask.sum()); o["box"] = box
        if tta:
            o["tta"] = [o["z_norm"], read_point(be.map(f, fr, flip=True), box, mask, ref, far)["z_norm"]]
            if "secondary" in be.variants:
                o["tta"].append(read_point(be.map(f, fr, secondary=True), box, mask, ref, far)["z_norm"])
        return o
    rows = []
    for r in PT.itertuples():
        o = at(int(r.frame), True); vec = [o["z_norm"]]
        for g in (r.frame - 2, r.frame - 1, r.frame + 1, r.frame + 2):
            if g in t.index and not bool(t.loc[g].es_puente):
                vec.append(at(int(g), False)["z_norm"])
        rows.append(dict(punto=r.punto, frame=r.frame, t_s=r.t_s, segment_id=r.segment_id, x_norm=r.x_norm, y_norm=r.y_norm, z_ave=o["z_ave"], z_flor=o["z_flor"], z_fondo=o["z_fondo"],
                         z_norm=float(np.mean(o["tta"])), z_norm_std=float(np.std(o["tta"])), z_norm_orig=o["z_norm"], vecinos_std=float(np.std(vec)), n_vecinos=len(vec) - 1,
                         metodo_z=o["metodo_z"], n_px_mascara=o["n_px"], den=o["den"], box=o["box"]))
    return pd.DataFrame(rows)


def control_ruido(D):
    """Regla fijada: la forma en z no es fiable si el ruido entre vecinos supera la diferencia típica entre puntos."""
    dif = float(np.median(np.abs(np.diff(D.z_norm)))); vec = float(D.vecinos_std.median()); tta = float(D.z_norm_std.median())
    return dict(dif_entre_puntos=dif, rango_z=float(np.ptp(D.z_norm)), ruido_vecinos=vec, ruido_tta=tta, forma_supera_ruido=bool(dif > vec),
                recortes=int((D.metodo_z != "mask_med").sum()), min_den=float(D.den.abs().min()))


def figura_3d(D, titulo, ref_rect, ref_nombre, AR, salida, xlim=None, ylim=None):
    """Tres vistas (jpg) y figura interactiva (html). x e y en unidades del ancho del cuadro; z = z_norm sin reescalar.
    La referencia se dibuja como un cilindro con el radio de la mitad del ancho de su región (supuesto)."""
    import plotly.graph_objects as go
    rr = ref_rect; P3 = np.c_[D.x_norm, D.y_norm * AR, D.z_norm]; cuts = np.flatnonzero(np.diff(D.segment_id.to_numpy()) != 0) + 1
    C3 = [catmull_rom(p) for p in np.split(P3, cuts)]
    cx, r = (rr[0] + rr[2]) / 2, (rr[2] - rr[0]) / 2; yt, yb = rr[1] * AR, rr[3] * AR
    fig, ax = plt.subplots(1, 3, figsize=(18, 5.2)); e = D.z_norm_std
    for a, i, j, tt in ((ax[0], 0, 1, "frontal x-y"), (ax[1], 0, 2, "superior x-z"), (ax[2], 2, 1, "lateral z-y")):
        for c in C3: a.plot(c[:, i], c[:, j], color="black", lw=2.2)
        a.errorbar(P3[:, i], P3[:, j], xerr=e if i == 2 else None, yerr=e if j == 2 else None, fmt="o", color="crimson", mec="w", capsize=3, zorder=5)
        for k, q in enumerate(P3, 1): a.annotate(str(k), (q[i], q[j]), xytext=(6, 5), textcoords="offset points", fontsize=11, weight="bold")
        a.set_title(tt, fontsize=11); a.grid(alpha=0.3)
    ax[0].add_patch(plt.Rectangle((rr[0], yt), 2 * r, yb - yt, fc="gold", alpha=0.35)); ax[0].set_aspect("equal")
    ax[1].add_patch(plt.Circle((cx, 0), r, fc="gold", alpha=0.35)); ax[1].axhline(0, color="goldenrod", ls=":"); ax[1].set_aspect("equal", adjustable="datalim")
    ax[2].add_patch(plt.Rectangle((-r, yt), 2 * r, yb - yt, fc="gold", alpha=0.35)); ax[2].axvline(0, color="goldenrod", ls=":")
    if xlim:
        ax[0].set_xlim(*xlim); ax[1].set_xlim(*xlim)
    if ylim:
        ax[0].set_ylim(*ylim); ax[2].set_ylim(*ylim)
    else:
        ax[0].autoscale_view(); ax[2].autoscale_view(); ax[0].invert_yaxis(); ax[2].invert_yaxis()
    fig.suptitle(f"{titulo} | z_norm: 0 = {ref_nombre}, 1 = fondo", fontsize=12); fig.tight_layout(); fig.savefig(f"{salida}.jpg", dpi=100); plt.show(); plt.close(fig)
    g = go.Figure(); th = np.linspace(0, 2 * np.pi, 48); T, YY = np.meshgrid(th, [yt, yb])
    g.add_trace(go.Surface(x=cx + r * np.cos(T), y=r * np.sin(T), z=-YY, colorscale=[[0, "gold"], [1, "gold"]], showscale=False, opacity=0.4, name=ref_nombre, showlegend=True))
    for c in C3: g.add_trace(go.Scatter3d(x=c[:, 0], y=c[:, 2], z=-c[:, 1], mode="lines", line=dict(color="black", width=7), name="curva"))
    g.add_trace(go.Scatter3d(x=P3[:, 0], y=P3[:, 2], z=-P3[:, 1], mode="markers+text", marker=dict(size=5, color="crimson"), text=[str(i) for i in D.punto], name="7 puntos"))
    g.update_layout(title=titulo, scene=dict(xaxis_title="x", yaxis_title="z_norm", zaxis_title="−y", aspectmode="data")); g.write_html(f"{salida}.html", include_plotlyjs=True)


def tira_puntos(D, cap, ref_rect, W, H, salida):
    """Recorte de cada punto con su z_norm, para revisar a ojo si la lectura tiene sentido."""
    x0, y0, x1, y1 = rect_px(ref_rect, W, H)
    fig, axes = plt.subplots(1, len(D), figsize=(3.2 * len(D), 3.3))
    for ax, r in zip(axes, D.itertuples()):
        fr = read_frame(cap, r.frame); b = r.box; cv2.rectangle(fr, (b[0], b[1]), (b[2], b[3]), (0, 0, 255), 2)
        X0, Y0, X1, Y1 = max(0, min(b[0], x0) - 40), max(0, min(b[1], y0) - 40), min(W, max(b[2], x1) + 40), min(H, max(b[3], y1) + 40)
        ax.imshow(cv2.cvtColor(fr[Y0:Y1, X0:X1], cv2.COLOR_BGR2RGB)); ax.set_xticks([]); ax.set_yticks([]); ax.set_xlabel(f"{r.punto} | f{r.frame} | {r.z_norm:+.2f} ± {r.z_norm_std:.2f}", fontsize=10)
    fig.tight_layout(); fig.savefig(salida, dpi=100); plt.show(); plt.close(fig)


# Portado de video-val/resultados/profundidad/siete_puntos/vda_inferir.py
def _leer_video_vda(path, short):
    cap = cv2.VideoCapture(str(path)); fps = cap.get(cv2.CAP_PROP_FPS); out = []
    while True:
        ok, f = cap.read()
        if not ok: break
        s = short / min(f.shape[:2])
        out.append(cv2.cvtColor(cv2.resize(f, None, fx=s, fy=s, interpolation=cv2.INTER_AREA), cv2.COLOR_BGR2RGB))
    cap.release()
    return np.stack(out), fps


def mapas_vda(videos_vda, cache, sizes=None):
    """Mapas de VDA Small de {nombre: ruta}, original y volteado, en <cache>/<nombre>_{orig,flip}.npz. Los que existen no se rehacen."""
    import gc, time
    import torch
    sizes = sizes or CFG["vda"]["sizes"]
    cache = Path(cache); cache.mkdir(parents=True, exist_ok=True)
    pendientes = [(n, p, t) for n, p in videos_vda.items() for t in ("orig", "flip") if not (cache / f"{n}_{t}.npz").exists()]
    if not pendientes:
        return
    repo, ckpt = ZONA / CFG["vda"]["repo"], ZONA / CFG["vda"]["checkpoint"]
    if not ckpt.exists():
        raise FileNotFoundError(f"Falta el checkpoint de VDA Small en {ckpt}: ver entornos/instalar.md, sección 3.")
    if str(repo) not in sys.path:
        sys.path.insert(0, str(repo))
    cwd = os.getcwd(); os.chdir(repo)
    from video_depth_anything.video_depth import VideoDepthAnything
    os.chdir(cwd)
    gc.collect(); torch.cuda.empty_cache()
    m = VideoDepthAnything(encoder="vits", features=64, out_channels=[48, 96, 192, 384])
    m.load_state_dict(torch.load(ckpt, map_location="cpu", weights_only=True), strict=True)
    m = m.to("cuda").eval()
    used = {}
    for name, path, tag in pendientes:
        for size in sizes:
            try:
                t0 = time.time(); torch.cuda.empty_cache(); torch.cuda.reset_peak_memory_stats()
                frames, fps = _leer_video_vda(path, size)
                if tag == "flip": frames = frames[:, :, ::-1].copy()
                d, _ = m.infer_video_depth(frames, fps, input_size=size, device="cuda", fp32=False)
                break
            except (torch.OutOfMemoryError, MemoryError) as e:
                print(name, tag, "sin memoria a", size, type(e).__name__, flush=True); frames = d = None; gc.collect(); torch.cuda.empty_cache()
        else:
            raise RuntimeError(f"{name}: no cabe en memoria a ninguna resolución; prueba con un clip más corto")
        d = np.asarray(d, dtype=np.float32)
        if tag == "flip": d = d[:, :, ::-1]
        np.savez_compressed(cache / f"{name}_{tag}.npz", maps=d.astype(np.float16))
        used[f"{name}_{tag}"] = size
        print(name, tag, d.shape, f"input {size}", f"{len(d) / (time.time() - t0):.1f} fps", f"pico GPU {torch.cuda.max_memory_allocated() / 1e9:.2f} GB", "rango", float(d.min()), float(d.max()), flush=True)
        frames = d = None; gc.collect(); torch.cuda.empty_cache()
    del m; gc.collect(); torch.cuda.empty_cache()
    previo = json.loads((cache / "info.json").read_text(encoding="utf-8")).get("input_size", {}) if (cache / "info.json").exists() else {}
    (cache / "info.json").write_text(json.dumps(dict(modelo="vda-small", input_size={**previo, **used}, salida="disparidad relativa, mayor = más cerca"), indent=2), encoding="utf-8")
````

## 9. Herramienta: videos y datasets propios

Funciones de `herramienta_trayectorias`. Reutilizan las secciones 5 a 8: el
tracker, la unión y los 7 puntos son los vigentes de `config.yaml`, sin verdad de terreno ni comparación con los sprints.
Las salidas van a `runs/herramienta/<video>/`.


````python
EXT_VIDEO = {".mp4", ".mov", ".avi", ".mkv", ".webm", ".m4v"}

def descargar_urls(urls, carpeta):
    """Baja cada URL a la carpeta con yt-dlp (YouTube, Vimeo, enlaces directos...). Lo ya descargado no se repite."""
    if not urls:
        return
    import subprocess
    carpeta = Path(carpeta); carpeta.mkdir(parents=True, exist_ok=True)
    for u in urls:
        subprocess.run([sys.executable, "-m", "yt_dlp", "--no-playlist", "--restrict-filenames", "-f", "bv*[ext=mp4]/b[ext=mp4]/b",   # solo video: no hace falta audio ni ffmpeg
                        "-o", str(carpeta / "%(title).60s_%(id)s.%(ext)s"), u], check=True)

def leer_fuente(fuente, archivos=None) -> dict:
    """{nombre: dict(path, W, H, fps, n)} de un video, de una carpeta (sin subcarpetas) o de `archivos` dentro de ella."""
    fuente = Path(fuente)
    if not fuente.exists():
        raise FileNotFoundError(f"No existe {fuente}. Crea la carpeta y pon ahí tus videos, o indica otra en CARPETA_VIDEOS.")
    if archivos:
        rutas = [fuente / a for a in archivos]
    elif fuente.is_file():
        rutas = [fuente]
    else:
        rutas = sorted(p for p in fuente.iterdir() if p.suffix.lower() in EXT_VIDEO)
    if not rutas:
        raise FileNotFoundError(f"No hay videos en {fuente} (formatos: {', '.join(sorted(EXT_VIDEO))}).")
    out = {}
    for p in rutas:
        cap = cv2.VideoCapture(str(p))
        if not cap.isOpened():
            print(f"✗ no se puede abrir {p.name}; se omite"); continue
        nombre = slug(p.stem)
        if nombre in out:
            nombre = f"{nombre}_{hashlib.sha1(p.name.encode()).hexdigest()[:6]}"
        out[nombre] = dict(path=p, W=int(cap.get(3)), H=int(cap.get(4)), fps=cap.get(cv2.CAP_PROP_FPS), n=int(cap.get(cv2.CAP_PROP_FRAME_COUNT)))
        cap.release()
    return out

def carpeta_salida(nombre):
    return P["herramienta"] / nombre

def procesar_video(nombre, v, detector=None):
    """Video -> tracking -> unión -> trayectoria -> 7 puntos, con la configuración vigente. Devuelve una fila descriptiva.
    Las detecciones se guardan por detector y no se recalculan ni se sobrescriben."""
    detector = Path(detector or DETECTOR); d = carpeta_salida(nombre); d.mkdir(parents=True, exist_ok=True)
    cfg_name = CFG["vigente"]["tracker"]; _, tracker, over = CONFIGS[cfg_name]
    det = d / f"tracks__{detector.parent.name}.csv"
    if det.exists():
        tr = pd.read_csv(det); tr.attrs.update(json.loads(det.with_suffix(".meta.json").read_text(encoding="utf-8")))
    else:
        tr = track_video(detector, v["path"], tracker, tracker_cfg(tracker, cfg_name, over))
        tr.to_csv(det, index=False); det.with_suffix(".meta.json").write_text(json.dumps(tr.attrs, indent=2), encoding="utf-8")
    traj = union_vigente(tr, scene_cuts(v["path"]))
    fila = dict(video=nombre, detector=detector.parent.name, duracion_s=round(v["n"] / v["fps"], 1), ids=int(tr.track_id.nunique()),
                segmentos=int(traj.segment_id.nunique()) if len(traj) else 0, tiempo_con_ave_s=round(len(traj) / v["fps"], 1),
                pct_con_ave=round(100 * len(traj) / max(v["n"], 1), 1), conf_mediana=round(float(tr.conf.median()), 3) if len(tr) else None, siete_puntos=False)
    (d / "trayectoria.csv").unlink(missing_ok=True)      # lo que queda en la carpeta corresponde a este detector
    if len(traj) < N:
        print(f"✗ {nombre}: {len(traj)} frames con ave, no alcanzan para {N} puntos. ¿Sale un colibrí? Prueba otro video o detector.")
        return fila
    traj = traj.sort_values("frame").reset_index(drop=True); traj.to_csv(d / "trayectoria.csv", index=False)
    br, largos = bridges(traj, v["W"], v["H"]); fila.update(puentes=len(br), puentes_largos=largos)
    draw(v["path"], traj, pd.DataFrame(dict(boxes=[])), f"{nombre} | {fila['segmentos']} segmentos | {fila['pct_con_ave']} % de frames con ave", d / "trayectoria.png")
    sp = siete_puntos(traj, v["fps"])
    cap = cv2.VideoCapture(str(v["path"])); cap.set(cv2.CAP_PROP_POS_FRAMES, int(traj.frame.median())); _, fr = cap.read(); cap.release()
    bg = cv2.cvtColor(fr, cv2.COLOR_BGR2RGB)
    for m, r in sp["metodos"].items():
        tabla_puntos(r, v["W"], v["H"]).to_csv(d / f"puntos_{m}.csv", index=False)
        lamina_fondo(sp, r, bg, v["W"], v["H"], f"{nombre} | {METHODS[m]} | {len(r['gaps'])} huecos | error medio {r['e_mean']:.0f} px", d / f"{m}_fondo.png")
        lamina_trazo(r, d / f"{m}_trazo.png")
        fila[f"error_px_{m}"] = round(r["e_mean"], 1)
    fila["siete_puntos"] = True
    return fila

def ver_cuadro(nombre, v, divisiones=10):
    """Muestra el frame mediano de la trayectoria (o del video) con una rejilla de 0 a 1, para escribir REGIONES."""
    f = carpeta_salida(nombre) / "trayectoria.csv"
    i = int(pd.read_csv(f).frame.median()) if f.exists() else v["n"] // 2
    cap = cv2.VideoCapture(str(v["path"])); fr = read_frame(cap, i); cap.release()
    fig, ax = plt.subplots(figsize=(10, 10 * v["H"] / v["W"])); ax.imshow(cv2.cvtColor(fr, cv2.COLOR_BGR2RGB), extent=(0, 1, 1, 0), aspect="auto")
    ax.set_xticks(np.linspace(0, 1, divisiones + 1)); ax.set_yticks(np.linspace(0, 1, divisiones + 1)); ax.grid(color="white", alpha=0.6, lw=0.8)
    ax.set_title(f"{nombre} | frame {i}", fontsize=10); plt.show(); plt.close(fig)

def procesar_3d(nombre, v, regiones, cache, metodos=("a_uniforme_tiempo", "b_salientes")):
    """Lee z en los 7 puntos con los mapas de VDA y dibuja la trayectoria 3D. regiones: {"flor": (nombre, rect), "fondo": (nombre, rect)}."""
    d = carpeta_salida(nombre); t = pd.read_csv(d / "trayectoria.csv").set_index("frame")
    cap = cv2.VideoCapture(str(v["path"])); W, H = v["W"], v["H"]
    bg = background(cap, v["n"]); be = make_backend("vda", "vda-small", nombre, (H, W), cache_dir=cache)
    (rn, rr), (_, ff) = regiones["flor"], regiones["fondo"]
    ref, far = rect_px(rr, W, H), rect_px(ff, W, H); filas = []
    for m in metodos:
        D = leer_puntos_z(pd.read_csv(d / f"puntos_{m}.csv"), t, cap, be, bg, ref, far, W, H)
        D.drop(columns=["box"]).to_csv(d / f"puntos3d_{m}.csv", index=False)
        figura_3d(D, f"{nombre} | {m} | vda-small", rr, rn, H / W, d / f"trayectoria3d_{m}")
        tira_puntos(D, cap, rr, W, H, d / f"tira_{m}.jpg")
        filas.append(dict(video=nombre, metodo=m, **control_ruido(D)))
    cap.release()
    return filas
````
